```python
import pandas as pd
import numpy as np

# 1. 创建一个"脏"数据集（故意放几个缺失值）
data = {
    "name": ["张三", "李四", "王五", "赵六", "孙七"],
    "math": [90, np.nan, 95, 60, 88],      # 李四的数学缺了
    "eng": [85, 92, np.nan, 70, 85],       # 王五的英语缺了
    "cprgm": [88, 80, 94, 65, np.nan]     # 孙七的C语言缺了
}
df = pd.DataFrame(data)
print("=== 原始数据（有缺失） ===")
print(df)
print()

# 2. 体检：看每列缺多少
print("=== 缺失值统计 ===")
print(df.isnull().sum())
print()

# 3. 方案A：直接删除有缺失的行（简单粗暴）
df_clean = df.dropna()
print("=== 删除缺失行后 ===")
print(df_clean)
print(f"剩余行数：{len(df_clean)}（原来5行，删了3行）")
print()

# 4. 方案B：用均值填充（保留所有样本）
df_fill = df.copy()  # 复制一份，不要动原来的 df
df_fill["math"] = df_fill["math"].fillna(df_fill["math"].mean())
df_fill["eng"] = df_fill["eng"].fillna(df_fill["eng"].mean())
df_fill["cprgm"] = df_fill["cprgm"].fillna(df_fill["cprgm"].mean())

print("=== 均值填充后 ===")
print(df_fill)
print(f"行数：{len(df_fill)}（还是5行，没有删）")
print()

# 5. 验证：填充后还有缺失吗？
print("=== 填充后缺失值统计 ===")
print(df_fill.isnull().sum())
```

```python
import numpy as np
import pandas as pd

data={
    "num":[1001,1002,1003,1004,1005,1006,1007,1008,1009,1010],
    "name":["张三","李四","王五","赵六","孙七","周八","吴九","郑十","钱十一","冯十二"],
    "math":[90,np.nan,95,60,88,92,75,85,70,98],
    "eng":[85,92,90,70,85,88,np.nan,95,65,92],
    "cprgm":[88,80,94,65,92,85,78,90,np.nan,95],
    "gender":["M","F","M","F","M","F","M","F","M","F"]
}
df=pd.DataFrame(data)
df.to_csv("students.csv",index=False,encoding="utf-8")
print("数据已保存到students.csv")
print(df)
```

```python
df= pd.read_csv("students.csv",encoding="utf-8")
print(f"总行数：{len(df)}")
print("\n前5行：")
print(df.head(5))
print("\n缺失值统计：")
print(df.isnull().sum())
```

```python
df["math"]=df["math"].fillna(df["math"].mean())
df["eng"]=df["eng"].fillna(df["eng"].mean())
df["cprgm"]=df["cprgm"].fillna(df["cprgm"].mean())

print("填充后缺失值统计:")
print(df.isnull().sum())
print("\n填充后的数据:")
print(df)
```

```python
df["total"]=df["math"]+df["eng"]+df["cprgm"]
df["avg"]=df["total"]/3.0
print(df[["name","math","eng","cprgm","total","avg"]])
```

```python
excellent=df[df["avg"]>85]
excellent.to_csv("excellent.csv",index=False,encoding="utf-8")
print(f"优秀生人数:{len(excellent)}")
print(excellent[["name","avg"]])
```

```python
gender_avg=df.groupby("gender")["total"].mean()
print("按性别分组的平均总分:")
print(gender_avg)
```

```python
math_mean=df["math"].mean()
eng_mean=df["eng"].mean()
cprgm_mean=df["cprgm"].mean()

print(f"数学平均分：{math_mean:.2f}")
print(f"英语平均分：{eng_mean:.2f}")
print(f"C语言平均分：{cprgm_mean:.2f}")

weak_math=df[df["math"]<math_mean-10]
weak_eng=df[df["eng"]<eng_mean-10]
weak_cprgm=df[df["cprgm"]<cprgm_mean-10]

weak_students = pd.concat([weak_math,weak_eng,weak_cprgm]).drop_duplicates()
print(f"\n偏科学生人数：{len(weak_students)}")
print(weak_students[["name","math","eng","cprgm"]])
```

```python
with open("report.txt", "w", encoding="utf-8") as f:
    f.write("=== 学生成绩分析报告 ===\n\n")
    f.write(f"总人数：{len(df)}\n")
    f.write(f"数学平均分：{math_mean:.2f}\n")
    f.write(f"英语平均分：{eng_mean:.2f}\n")
    f.write(f"C语言平均分：{cprgm_mean:.2f}\n\n")
    
    f.write(f"优秀生（平均分>85）：{len(excellent)}人\n")
    for _, row in excellent.iterrows():
        f.write(f"  {row['name']}: {row['avg']:.2f}分\n")
    
    f.write(f"\n性别分组平均总分：\n")
    f.write(f"  男生：{gender_avg['M']:.2f}\n")
    f.write(f"  女生：{gender_avg['F']:.2f}\n")
    
    f.write(f"\n偏科学生：{len(weak_students)}人\n")
    for _, row in weak_students.iterrows():
        f.write(f"  {row['name']}\n")

print("报告已保存到 report.txt")
```
