```python
import numpy as np
from collections import Counter

print("环境OK")
```

```python
docs=[
    "机器学习 是 人工智能 的 一个 分支",
    "深度学习 是 机器学习 的 一个 子集",
    "自然语言处理 是 人工智能 的 应用"]

tokenized=[d.split()for d in docs]
print("分词结果:",tokenized)

vocab=sorted(set(w for doc in tokenized for w in doc))
print("词表:",vocab)

def tf(doc,vocab):
    counts=Counter(doc)
    total=len(doc)
    return np.array([counts.get(w,0)/total for w in vocab])

N = len(docs)
idf=[]
for w in vocab:
    df=sum(1 for doc in tokenized if w in doc)
    idf.append(np.log(N/df))
idf=np.array(idf)

print("\n每个词的IDF:")
for w,v in zip(vocab,idf):
    print(f"  {w}:{v:.4f}")

tfidf=[]
for doc in tokenized:
    t = tf(doc,vocab)
    tfidf.append(t*idf)

tfidf = np.array(tfidf)
print("\nTF-IDf矩阵:")
print(np.round(tfidf,4))
```
