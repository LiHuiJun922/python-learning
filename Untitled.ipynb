```python
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score

# ---------------------- 1. 造一份二分类toy数据集 ----------------------
np.random.seed(42)
n = 200
# 两类样本
X0 = np.random.randn(n//2,2) + np.array([-2,-2])
X1 = np.random.randn(n//2,2) + np.array([2,2])
X_raw = np.vstack([X0,X1])
y = np.hstack([np.zeros(n//2), np.ones(n//2)]).reshape(-1,1)

# 增加偏置项，构造 X_b (n_samples, feature+1)
n_samples = X_raw.shape[0]
X_b = np.hstack([np.ones((n_samples,1)), X_raw])

# ---------------------- 2. sigmoid 和交叉熵损失 ----------------------
def sigmoid(z):
    return 1.0/(1.0+np.exp(-z))

def compute_loss(X, y, w):
    m = X.shape[0]
    z = X @ w
    y_prob = sigmoid(z)
    eps = 1e-8
    loss = (-1/m) * np.sum( y*np.log(y_prob+eps) + (1-y)*np.log(1-y_prob+eps) )
    return loss

# ---------------------- 3. 超参数 ----------------------
lr = 0.5
epochs = 1000

# ---------------------- 4. StratifiedKFold 5折交叉验证 ----------------------
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
f1_scores = []

for fold_idx, (train_idx, val_idx) in enumerate(kf.split(X_b, y)):
    print(f"\n==========第 {fold_idx+1} 折==========")
    X_train, X_val = X_b[train_idx], X_b[val_idx]
    y_train, y_val = y[train_idx], y[val_idx]
    
    w = np.zeros((X_train.shape[1], 1))
    m_train = X_train.shape[0]
    loss_history = []
    
    print("训练集标签计数：", np.unique(y_train.ravel(), return_counts=True))
    print("验证集标签计数：", np.unique(y_val.ravel(), return_counts=True))

    # 梯度下降训练
    for i in range(epochs):
        z = X_train @ w
        y_prob = sigmoid(z)
        loss = compute_loss(X_train, y_train, w)
        loss_history.append(loss)
        grad = (1/m_train) * X_train.T @ (y_prob - y_train)
        w = w - lr * grad

    w_norm = np.linalg.norm(w)
    final_loss = loss_history[-1]
    print(f"w的L2范数: {w_norm:.4f}, 最终loss: {final_loss:.4f}")

    # 预测评估
    val_prob = sigmoid(X_val @ w)
    y_pred = (val_prob >= 0.5).astype(int)
    
    print("验证集前6个概率：", val_prob.ravel()[:6])
    print("验证集真实标签：", y_val.ravel()[:6])
    print("验证集预测标签：", y_pred.ravel()[:6])

    f1 = f1_score(y_val.ravel(), y_pred.ravel())
    f1_scores.append(f1)
    print(f"本折F1 = {f1:.4f}")


print("\n======= 5折交叉验证最终结果 =======")
print(f"5折F1列表: {np.round(f1_scores,4)}")
print(f"平均F1: {np.mean(f1_scores):.4f}")
```
