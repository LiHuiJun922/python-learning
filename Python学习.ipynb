```python
##定义函数 calc(math, eng, cprgm)，计算总分和平均分，返回两个值。
##创建列表 students，包含两个字典，每个字典有键：num、math、eng、cprgm，值随意。
##用 for 遍历 students，调用 calc 计算平均分，以 "avg" 为键存入字典。
##对 students 按 "avg" 降序排序，结果存回 students。
##遍历排序后的列表，按 学号 数学 英语 C语言 平均分 格式打印，平均分保留两位小数。
def calc(math,eng,cprgm):
    total=math+eng+cprgm
    avg=(math+eng+cprgm)/3
    return total,avg
students = [{"num": 101, "math": 90, "eng": 85, "cprgm": 88},
    {"num": 102, "math": 78, "eng": 92, "cprgm": 80},]
for s in students:
    s["avg"]=(s["math"]+s["eng"]+s["cprgm"])/3
sorted(students,key=lambda x:avg,reverse=True)
for s in students:
    print(f"{s['num']} {s['math']} {s['eng']} {s['cprgm']} {s['avg']:.2f}" )


```
