```python
import pandas as pd
data ={
    "num":[1001,1002,1003,1004,1005],
    "name":["张三","李四","王五","赵六","孙七"],
    "math":[90,78,95,60,88],
    "eng":[85,92,90,70,85],
    "cpgrm":[88,80,94,65,92],
    "gender":["M","F","M","F","M"]
}
df=pd.DataFrame(data)
df.to_csv("score.csv",index=False,encoding="utf-8")
print("文件已创建")

```

```python
import pandas as pd
df=pd.read_csv("score.csv",encoding="utf-8")
print(df.head())
print(df.info())
print(df.describe())
```

```python
import pandas as pd
df=pd.read_csv("score.csv",encoding="utf-8")
print(df[df["math"]>=90])
print(df[(df["math"]>=85)&(df["eng"]>=90)])
```

```python
import pandas as pd
df=pd.read_csv("score.csv",encoding="utf-8")
print(df.groupby("gender")["math"].mean())
```

```python
import pandas as pd
df=pd.read_csv("score.csv",encoding="utf-8")
df["total"]=df["math"]+df["eng"]+df["cpgrm"]
df["avg"]=df["total"]/3.0
df_sorted=df.sort_values("avg",ascending=False)
print(df_sorted)
```

```python
import pandas as pd
df=pd.read_csv("score.csv",encoding="utf-8")
df_sorted.to_csv("score_sorted.csv",index=False,encoding="utf-8")
print("排序结果已保存")
```
